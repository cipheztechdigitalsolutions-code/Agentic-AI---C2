```python
from dotenv import load_dotenv
import os
load_dotenv(override=True)
```

```python
from crewai import LLM, Agent, Task, Crew

llm = LLM(
    model="openrouter/openai/gpt-4o-mini",
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("API_TOKEN")
)
```

```python
researcher = Agent(
    role="Senior Research Analyst",
    goal="Find and Analyze the latest AI trends",
    backstory="You are a seasoned research analyst with 10 years of experience in technology research. You are known for thorough analysis and ability to spot emerging trends before they become mainstream",
    llm=llm,
    verbose=False
)
```

```python
task = Task(
    title="Research",
    description="Perform a research on the history of gemini",
    expected_output="I need a detailed report on gemini",
    agent=researcher
)
```

```python
crew = Crew(
    agents=[researcher],
    tasks=[task],
    verbose=False
)
```

```python
result = await crew.kickoff_async()
print(result)
```

CrewAI Tools


```python
from crewai.tools import tool, BaseTool
```

```python
@tool("Calculator")
def calculator(expression: str) -> str:
    """Evaluate a mathematical expression and return the result."""
    try:
        return f"Good Calculations: {str(eval(expression))}"
    except Exception as e:
        return f"Error: {e}"
```

```python
class StockPriceTool(BaseTool):
    name: str = "Stock Price Checker"                           # Tool name
    description: str = "Gets the current stock price for a given ticker symbol. Use when asked about stock prices."  # When to use it
    
    def _run(self, ticker: str) -> str:
        """
        Args:
            ticker: Stock ticker symbol like AAPL, GOOGL, MSFT
        """
        # In a real app, you'd call a financial API here
        # For demo, return mock data
        prices = {
            "AAPL": 195.50,
            "GOOGL": 175.20,
            "MSFT": 420.00,
            "TSLA": 245.80
        }
        
        price = prices.get(ticker.upper())
        if price:
            return f"{ticker.upper()} is currently trading at ${price}"
        else:
            return f"Could not find price for {ticker}"

```

How you pass tools to the Agent Component created from Tool Decorator

```python
math_agent = Agent(
    role="Math Assistant",
    goal="Help Users solve their problem accurately",
    backstory="You are a Math Professor that solves people's problem ......, When it comes to scraping websites to get more information, pls use the calculator tool",
    llm=llm,
    tools=[calculator],
    verbose=False

)
```

How you pass tools to the Agent Component created from BaseTool

```python
math_agent = Agent(
    role="Math Assistant",
    goal="Help Users solve their problem accurately",
    backstory="You are a Math Professor that solves people's problem ......, When it comes to scraping websites to get more information, pls use the calculator tool",
    llm=llm,
    tools=[StockPriceTool()],
    verbose=False

)
```

```python
task = Task(
    title="Mathemactical Problem",
    description="I need you to solve a multiplication problem: {topic}",
    expected_output="I will need a simple response as an answer with brief explanation",
    agent=math_agent
)
```

```python
crew = Crew(
    agents=[math_agent],
    tasks=[task],
    verbose=False
)
```

```python
result = await crew.kickoff_async()
print(result)
```

```python
result = await crew.kickoff_async(
    inputs={"topic": "Multiply 2 and 3"}
)
print(result)
```
